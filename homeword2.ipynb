```python
# ax^2+bx+c=0
a, b, c = map(float, input("Nhập a, b và c: ").split())
delta = b**2 - 4*a*c
if delta < 0:
    print("Phương trình vô nghiệm")
elif delta == 0:
    x = -b / (2*a)
    print("Phương trình có nghiệm kép: ", x)
else:
    x1 = (-b + delta**0.5) / (2*a)
    x2 = (-b - delta**0.5) / (2*a)
    print("Phương trình có hai nghiệm phân biệt: ", x1, "và", x2)
```

```python
a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))
d = float(input("d = "))
m = float(input("m = "))
n = float(input("n = "))

D = a*d - b*c

if D != 0:
    x = (m*d - b*n) / D
    y = (a*n - m*c) / D
    print("x =", x)
    print("y =", y)
else:
    print("Hệ phương trình không có nghiệm duy nhất")
```

```python
a=float(input("Bán kính: "))
c,d=map(float,input("Tọa độ tâm: ").split())
x,y=map(float,input("Tọa độ điểm: ").split())
if (x-c)**2+(y-d)**2<a**2:
    print("Điểm nằm trong hình tròn")
elif (x-c)**2+(y-d)**2>a**2:
    print("Điểm nằm ngoài hình tròn")
else:
    print("Điểm nằm trên đường tròn")
```
